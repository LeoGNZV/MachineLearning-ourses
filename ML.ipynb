Остальные значения-пики нельзя считать выбросами, так как они возникают регулярно. Скорее всего, они напрямую связаны с начальным уровнем:

```python
bins = [0, 4, 8, 12, 16, 20]
labels = ['1–4', '5–8', '9–12', '13–16', '17–20']
data2 = data
data2['level_range'] = pd.cut(data2['class_starting_level'], bins=bins, labels=labels)

fig, axes = plt.subplots(len(labels), 1, figsize=(14, 5 * len(labels)),
                         sharex=True)

for ax, rng in zip(axes, labels):
    subset = data2[data2['level_range'] == rng]
    for cl in sorted(subset['class_starting'].dropna().unique()):
        sub = subset.loc[subset['class_starting'] == cl, 'base_hp']
        counts = (sub.value_counts()
                  .sort_index()
                  .reindex(range(0, 16), fill_value=0))
        ax.plot(counts.index, counts.values, marker='o', label=cl, alpha=0.7)

    ax.set_title(f'class_starting_level: {rng}')
    ax.set_ylabel('количество')
    ax.set_xticks(range(0, 16))
    ax.grid(alpha=0.3)

axes[-1].set_xlabel('base_hp')
axes[0].legend(bbox_to_anchor=(1.02, 1), loc='upper left')
plt.tight_layout()
plt.show()
```

```python
dummies_train = pd.DataFrame(enc.transform(X_train[cat_cols]),
                             columns=enc.get_feature_names_out(cat_cols),
                             index=X_train.index)
dummies_test  = pd.DataFrame(enc.transform(X_test[cat_cols]),
                             columns=enc.get_feature_names_out(cat_cols),
                             index=X_test.index)
```

```python
X_train = pd.concat([X_train.drop(columns=cat_cols), dummies_train], axis=1)
X_test  = pd.concat([X_test.drop(columns=cat_cols),  dummies_test],  axis=1)

X_test = X_test.reindex(columns=X_train.columns, fill_value=0)
```

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
scaler.fit(X_train)
X_train = pd.DataFrame(scaler.transform(X_train),
                       columns=X_train.columns, index=X_train.index)
X_test  = pd.DataFrame(scaler.transform(X_test),
                       columns=X_train.columns, index=X_test.index)
```

```python
from sklearn.metrics import accuracy_score, f1_score
from sklearn.linear_model import LogisticRegression

model = LogisticRegression(max_iter=1000, class_weight='balanced', n_jobs=-1)
model.fit(X_train, y_train)

results = []
for split, X_, y_ in [('train', X_train, y_train),
                      ('test',  X_test,  y_test)]:
    y_pred = model.predict(X_)
    results.append({
        'model': f"LogReg {split}",
        'accuracy': accuracy_score(y_, y_pred),
        'f1_macro': f1_score(y_, y_pred, average='macro', zero_division=0),
        'f1_weighted': f1_score(y_, y_pred, average='weighted', zero_division=0),
    })

print(pd.DataFrame(results[-2:]).round(3).to_string(index=False))
```

Получено достаточно хорошее предсказание, учитывая, что классов всего 14; переобучения, как видно из результатов, нет

Accuracy - количество успешных предсказаний из общего числа.
f1_macro - среднее количество f1 по всем классам
f1 строится для конкретного класса. Это 2 умножить на произведение полноты и точность, а также на сумму полноты и точности, где точность - количество точных угадываний класса к общему числу предположений о конкретном классе, а полнота - количество точных угадываний класса к сумме первого, а также ошибок модели, когда она предположила другой класс. f1_weighted - f1 взвешенное по частоте классов

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# --- коэффициенты логистической регрессии ---
coefs = model.coef_                     # форма: (n_classes, n_features)
print("Форма coef_:", coefs.shape)      # например (12, 57)

# --- важность = средний модуль коэффициента по всем классам ---
importances = pd.Series(
    np.abs(coefs).mean(axis=0),         # среднее |β| по классам
    index=X_train.columns               # имена признаков
).sort_values(ascending=False)

print("\nТоп-20 важных признаков (LogReg, средний |β|):")
print(importances.head(20).round(3))

# --- график ---
plt.figure(figsize=(10, 8))
sns.barplot(y=importances.head(20).index,
            x=importances.head(20),
            orient="h")
plt.title('LogisticRegression: важность признаков (средний |β|)')
plt.xlabel('средний |β| по классам')
plt.tight_layout()
plt.show()
```
